```python
import pandas as pd
df1 = pd.read_csv('/home/ushtro/Documents/AIML/fed_l/results/delay/delay_hyperparam_sweep_20250912_000655.csv')
df2 = pd.read_csv('/home/ushtro/Documents/AIML/fed_l/results/delay/delay_hyperparam_sweep_20250912_230155.csv')

df = pd.concat([df1, df2])
df.drop_duplicates(inplace=True)
df.head()
```

```python
# import matplotlib.pyplot as plt
# import seaborn as sns

# for col in df.columns:
#     plt.figure(figsize=(3,3))
#     if col == 'max_r2_delay':
#         break
#     sns.boxplot(data=df, x = col, y='max_r2_delay')
#     t = f"max_r2_delay vs {col}"
#     plt.title(t)
#     plt.show()

import matplotlib.pyplot as plt
import seaborn as sns

# Assuming df is your DataFrame
# and 'max_r2_delay' is one of the columns

cols = [col for col in df.columns if col != 'max_r2_delay']
n_cols = len(cols)

# Calculate the number of rows and columns for the subplot grid
# The number of columns is fixed at 3
n_rows = (n_cols + 2) // 3  # This ensures enough rows for all plots

fig, axes = plt.subplots(n_rows, 3, figsize=(10,10))

# Flatten the axes array to easily iterate through all subplots
axes = axes.flatten()

# Iterate through the columns and the axes simultaneously
for i, col in enumerate(cols):
    sns.boxplot(data=df, x=col, y='max_r2_delay', ax=axes[i])
    axes[i].set_title(f"max_r2_delay vs {col}")

# Hide any unused subplots
for j in range(n_cols, len(axes)):
    axes[j].axis('off')

plt.tight_layout()
plt.show()
```
