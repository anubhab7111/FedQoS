```python
import matplotlib.pyplot as plt
import numpy as np

sample_rate = 1000
duration = 2
N = sample_rate * duration

time = np.linspace(0, duration, N, endpoint=False)

signal_50hz = np.sin(50*2*np.pi*time)
signal_120hz = 0.5 * np.sin(120*2*np.pi*time)

signal = signal_120hz + signal_50hz

# fft
fft_raw = np.fft.fft(signal)
fft_mag = np.abs(fft_raw)
freq = np.fft.fftfreq(N, 1 / sample_rate)

# plot
n_half = N // 2
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(time, signal)
plt.title("Signal in Time Domain")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid()

# Plot the FFT result in the frequency domain
plt.subplot(1, 2, 2)
plt.plot(freq[:n_half], fft_mag[:n_half])
plt.title("Signal in Frequency Domain (FFT)")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.grid()

plt.tight_layout()
plt.show()
```
