```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a noisy signal for demonstration
time_steps = np.linspace(0, 100, 500)
original_signal = pd.Series(np.sin(time_steps / 5) * 10)
noisy_signal = original_signal + np.random.randn(500) * 3

window_size = 20
smoothed_signal = noisy_signal.rolling(window=window_size).mean()

# Plot the results
plt.figure(figsize=(15, 6))
plt.plot(noisy_signal, label='Noisy Signal', color='gray', alpha=0.6)
plt.plot(smoothed_signal, label=f'Moving Average (N={window_size})', color='red', linewidth=2)
plt.plot(original_signal, label='Original Signal', color='blue', linestyle='--', linewidth=2)
plt.title('Moving Average Filter Demonstration')
plt.legend()
plt.grid(True)
plt.show()
```
